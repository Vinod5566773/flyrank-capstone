## 1. Question

*The research question and the decision it supports.*

Can recent search-performance signals identify content experiencing meaningful traffic decline and help prioritize content for review or refresh?

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

This project uses the FlyRank ML Internship search-intelligence warehouse, accessed through the gated Hugging Face release. The analysis aggregates daily search-performance data into content-level 30-day windows. The final model and ranking dataset contains 111,247 content items with sufficient history, using a minimum of 100 impressions in the previous 30-day period. The main signals include previous-period impressions, visible-query activity, rare-query share, anonymous-query share, top-query share, and kept impressions. A 90-day feature-engineering exercise was also tested as a robustness check, but the reported model comparison uses the final 30-day feature table. Only aggregate results are presented in the public report; client names, domains, private queries, URLs, credentials, and raw exports are excluded.


## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

The analysis defines a declining content item as one where impressions in the most recent 30-day period are less than 80% of impressions in the preceding 30-day period. The main model uses previous-period impressions, visible-query activity, rare-query share, anonymous-query share, and top-query share as features. A second model adds kept impressions to test whether this signal improves performance on unseen clients. A 200-tree Random Forest classifier was evaluated using both a stratified random row split and a client-held-out split, with majority-class accuracy used as the baseline. For prioritization, an opportunity score combines traffic decline and normalized kept impressions, followed by rule-based action categories for investigation, refresh, monitoring, or maintenance. These scores and actions are intended as operational prioritization heuristics rather than causal claims.


## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

The random row split achieved 65.6% accuracy compared with a 63.3% majority-class baseline, with an F1 score of 0.756 for the declining class. On the stricter client-held-out split, accuracy decreased to 61.1% compared with a 67.7% baseline, while the declining-class F1 score was 0.704. Adding kept impressions improved the client-held-out result to 66.3% accuracy and 0.745 declining-class F1, although accuracy remained slightly below the majority baseline.

| Validation                         | Accuracy | Declining-class F1 | Baseline Accuracy |
| ---------------------------------- | -------: | -----------------: | ----------------: |
| Random row split                   |    65.6% |              0.756 |             63.3% |
| Client-held-out split              |    61.1% |              0.704 |             67.7% |
| Client-held-out + kept impressions |    66.3% |              0.745 |             67.7% |

The resulting rule-based prioritization categories contain 42,756 items for review or refresh, 39,700 for maintain/monitor, 27,631 for monitoring declining performance, and 1,160 for investigating traffic loss. These categories describe the output of the prioritization rules and should not be interpreted as evidence of causal business impact.


## 5. Limitations

*What this work cannot claim.*

The client-held-out evaluation shows that the reported model does not outperform the majority-class baseline on accuracy, so generalization to unseen clients is not established. Although adding kept impressions improved the declining-class F1 score and accuracy relative to the basic client-held-out model, the accuracy remained below the baseline. The opportunity score is a heuristic and can strongly prioritize content with zero recent impressions, which requires human review before action. The action categories are operational rules rather than causal recommendations. The analysis does not establish that any observed decline was caused by a search-engine algorithm change or that refreshing content would cause traffic to recover. The anonymized dataset also limits the contextual information available for diagnosing individual content items.


## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

The following operational priority order is derived from the rule-based opportunity score and decline thresholds:

1. **Investigate traffic loss** — prioritize content with zero impressions in the most recent 30-day period.
2. **Review or refresh content** — prioritize content with an observed impression decline of 50% or more.
3. **Monitor declining performance** — track content with an impression decline of 20% to 50%.
4. **Maintain / monitor** — continue monitoring content that does not meet the decline thresholds.

This ordering is an operational workflow for reviewing the dataset, not a model-proven ranking of business value. Individual items should be reviewed using additional context before any content changes are made.
