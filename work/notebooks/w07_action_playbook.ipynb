## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

### Ranked actions + reason codes

The ranked queue is a decision-support tool for prioritizing human review. It uses an opportunity score that combines observed impression decline and a low-kept-impressions component.

**Reason codes and suggested actions:**

* `strong_traffic_decline`: Investigate the traffic decline and check whether tracking, seasonality, or other changes explain it.
* `low_kept_impressions`: Review available evidence and decide whether the item merits further investigation.
* `monitor`: Keep the item under observation unless new evidence suggests action.

**Action priorities:**

1. Investigate items with zero latest-window impressions before deciding whether to refresh them.
2. Review items with substantial observed declines and sufficient previous-window visibility.
3. Monitor items with weaker or ambiguous signals.

The highest-ranked items in my current analysis all had zero latest-window impressions. Their scores were nearly identical, so the queue has limited ability to distinguish among these items. The ranking should guide investigation, not automatically determine which content gets refreshed.


```python
reason_codes = [
    "strong_traffic_decline",
    "low_kept_impressions",
    "monitor",
]

action_priorities = [
    "Investigate zero latest-window impressions",
    "Review substantial declines",
    "Monitor weaker or ambiguous signals",
]

for code, action in zip(reason_codes, action_priorities):
    print(f"{code}: {action}")
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human review + the no-go list

**Before acting, a human reviewer must:**

1. Confirm that the impression measurements and time windows are comparable.
2. Check for missing data, tracking changes, seasonality, and unusual traffic events.
3. Review the item's previous and latest impressions and the reason code.
4. Check whether the available query and position signals support the proposed action.
5. Record the reason for the decision and reassess after new evidence becomes available.

**No-go list — never automate these decisions:**

* Automatically refreshing, deleting, or rewriting content based only on the score.
* Treating the score as a probability that a refresh will succeed.
* Claiming that a content change or algorithm update caused the observed decline without suitable evidence.
* Publishing client-identifying information or private queries.
* Ignoring missing data or sending an item for action without human review.

The queue supports investigation and prioritization; the final decision remains with a human reviewer.


```python
review_checks = [
    "Verify measurement windows",
    "Check missing data and tracking changes",
    "Review impressions and reason code",
    "Check supporting signals",
    "Record the human decision",
]

no_go_actions = [
    "Automatic content refresh or deletion",
    "Treating scores as probabilities",
    "Unsupported causal claims",
    "Publishing private queries or identifying details",
]

print("Human review checklist:")
for item in review_checks:
    print("-", item)

print("\nNo-go actions:")
for item in no_go_actions:
    print("-", item)
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

### Monitoring / retrain triggers

The queue and model should be reviewed periodically to check whether their recommendations remain useful.

**Monitoring triggers:**

* A change in the definition or collection of impression measurements.
* A rise in missing values or changes in feature coverage.
* A substantial shift in impression or query-signal distributions.
* More top-ranked items receiving low-priority decisions from human reviewers.
* A growing gap between model performance and the majority-class baseline.

**Retraining and revalidation:**

* Revalidate after a material data-pipeline or feature-definition change.
* Retrain only when enough new, appropriately labelled data is available.
* Evaluate on a held-out client or time period and compare against the same baseline.
* Record the dataset window, feature list, split method, random seed, and evaluation metrics.

No monitoring thresholds have been established from production evidence yet. The triggers above are proposed checks, not measured production alerts. Revalidation should happen before the updated model is used for decisions.


```python
monitoring_triggers = [
    "Changes in measurement definitions",
    "Increasing missing-value rates",
    "Shifts in feature distributions",
    "More recommendations rejected by reviewers",
    "Model performance falling below the baseline",
]

for trigger in monitoring_triggers:
    print("-", trigger)

print("\nRetrain only after checking new labelled data and revalidating.")
print("Production thresholds: not yet established.")
```
