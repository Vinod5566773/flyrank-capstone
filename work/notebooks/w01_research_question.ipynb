## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This analysis can describe observed patterns in content impressions and trends. It can provide directional, decision-support recommendations about which content items may deserve review first. However, the analysis cannot prove that refreshing a page will cause its performance to improve. It also cannot predict Google's algorithm or guarantee future traffic. Recommendations should be reviewed by a person before action is taken.


```python
claims = [
    "Describe observed patterns in content performance.",
    "Provide directional decision support for content review."
]

limitations = [
    "Cannot prove that refreshing content causes improvement.",
    "Cannot predict Google's algorithm or guarantee future traffic."
]

assert len(claims) == 2
assert len(limitations) == 2

print("Supported claims:")
for claim in claims:
    print("-", claim)

print("\nLimitations:")
for limitation in limitations:
    print("-", limitation)
```
