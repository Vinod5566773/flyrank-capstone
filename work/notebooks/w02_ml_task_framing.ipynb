## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is **Refresh / Content Opportunity Scoring**. The prediction task is a classification problem because the model predicts whether a content item shows a defined decline in impressions. I use the prediction and observed performance signals to help prioritize items for review. The final recommendations are decision-support suggestions, not guarantees that refreshing a page will improve its performance.


```python
task_type = "Classification"
print("ML Task Type:", task_type)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is `is_declining`, a binary label defined from observed impressions. It is 1 when impressions in the most recent 30-day window are less than 80% of impressions in the preceding 30-day window, and 0 otherwise. This is a rule-based proxy for declining performance, not a human-confirmed label that a page needs refreshing. A decline does not prove that refreshing the content will improve performance.


```python
target_name = "is_declining"
target_rule = "imp_last30 < 0.8 * imp_prev30"
label_type = "Rule-based proxy from observed impressions"

print("Target:", target_name)
print("Label rule:", target_rule)
print("Label type:", label_type)
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will evaluate the classifier using class-1 F1 and accuracy, and compare its accuracy with the majority-class baseline. Class-1 F1 balances precision and recall for the declining class. I will also examine performance on a client-held-out split because a random row split may give an overly optimistic view of generalization. A model is useful only if its performance is interpreted against these baselines and limitations.


```python
metric = "Class-1 F1, with accuracy and majority-baseline comparison"
print("Success Metric:", metric)
```

The unit of analysis is one content item for a particular client. Each row in the modeling table represents a content item and contains aggregated performance features, including previous-window impressions, recent-window impressions, recent clicks, and other available signals. The target indicates whether the item meets the defined impression-decline rule.


```python
unit_of_analysis = "One content item for a client"

print("Unit of analysis:", unit_of_analysis)
print("Each row represents a content item with aggregated performance features.")
print("The target indicates whether impressions meet the defined decline rule.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A single rule based only on an impression drop may overlook differences between content items. Machine learning can combine several available performance signals to identify patterns associated with the defined decline label. However, the model must be compared with a simple baseline and tested on held-out clients. Its output helps prioritize reviews; it does not establish why performance changed or guarantee that a refresh will work.


```python
print("Reason:")
print(
    "Machine learning can combine multiple performance signals "
    "to identify patterns associated with declining impressions."
)
print(
    "The model supports review prioritization; it does not prove "
    "that refreshing content will improve performance."
)
```
